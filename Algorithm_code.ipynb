**Code of Merge Sort**

```
import random

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    # divedi list to half
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # merging
    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    # add smaller value
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # add rest value
    result.extend(left[i:])
    result.extend(right[j:])
    return result


# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[0:10])
sorted_arr = merge_sort(arr)
print("정렬된 배열: ", sorted_arr[:10])
```

**Code of Heap Sort**

```
import random

def heapify(arr, n, i):
  largest = i
  left = 2 * i + 1   # left index
  right = 2 * i + 2  # right index

  # left node is larger then largest
  if left < n and arr[left] > arr[largest]:
      largest = left

  # right node is larger then largest
  if right < n and arr[right] > arr[largest]:
      largest = right

  # largest == root
  if largest != i:
      arr[i], arr[largest] = arr[largest], arr[i]
      heapify(arr, n, largest)

def heap_sort(arr):
  n = len(arr)

  # make max heap
  for i in range(n // 2 - 1, -1, -1):
      heapify(arr, n, i)

  for i in range(n - 1, 0, -1):
      # exchange root and last value
      arr[i], arr[0] = arr[0], arr[i]
      heapify(arr, i, 0)

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
heap_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Bubble Sort**

```
import random

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        # check already sorted
        swapped = False
        for j in range(0, n - i - 1):
            # compare close two value
            if arr[j] > arr[j + 1]:
                # exchange
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
bubble_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Insertion Sort**

```
import random

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]           # insert value
        j = i - 1

        # move value larget then key to right
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        # insert key
        arr[j + 1] = key

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
insertion_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Selection Sort**

```
import random

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i  # i == minimum
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        # exchange minimum and present
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
selection_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Quick Sort**

```
import random

def quick_sort(arr):
    # stop recursion
    if len(arr) <= 1:
        return arr

    pivot = arr[0]  # make pivot
    left = [x for x in arr[1:] if x <= pivot]
    right = [x for x in arr[1:] if x > pivot]

    # sorting and merging by recursion
    return quick_sort(left) + [pivot] + quick_sort(right)

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
quick_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Library Sort**

```
import random
import bisect

def library_sort(arr, epsilon=0.5):
    if not arr:
        return []

    n = len(arr)
    size = int((1 + epsilon) * n) + 1
    result = [None] * size
    sorted_positions = []

    mid = size // 2
    result[mid] = arr[0]
    sorted_positions.append(mid)

    for i in range(1, n):
        val = arr[i]

        active_values = [result[pos] for pos in sorted_positions]
        idx = bisect.bisect_left(active_values, val)

        # calculate insert position
        if idx == 0:
            insert_pos = sorted_positions[0] - 2
        elif idx == len(sorted_positions):
            insert_pos = sorted_positions[-1] + 2
        else:
            insert_pos = (sorted_positions[idx - 1] + sorted_positions[idx]) // 2

        # find insert position
        attempts = 0
        while (
            insert_pos < 0 or insert_pos >= size
            or result[insert_pos] is not None
        ) and attempts < 10:
            insert_pos += 1
            attempts += 1

        if attempts >= 10:
            # re-arrange
            result, sorted_positions = rebalance(result, sorted_positions)
            size = len(result)
            # re-calculate insert position
            active_values = [result[pos] for pos in sorted_positions]
            idx = bisect.bisect_left(active_values, val)

            if idx == 0:
                insert_pos = sorted_positions[0] - 2
            elif idx == len(sorted_positions):
                insert_pos = sorted_positions[-1] + 2
            else:
                insert_pos = (sorted_positions[idx - 1] + sorted_positions[idx]) // 2

            while (
                insert_pos < 0 or insert_pos >= size
                or result[insert_pos] is not None
            ):
                insert_pos += 1

        result[insert_pos] = val
        bisect.insort(sorted_positions, insert_pos)

    return [x for x in result if x is not None]

def rebalance(array, positions):
    new_size = len(array) * 2
    new_array = [None] * new_size
    gap = new_size // (len(positions) + 1)
    new_positions = []

    for i, pos in enumerate(positions):
        new_pos = (i + 1) * gap
        new_array[new_pos] = array[pos]
        new_positions.append(new_pos)

    return new_array, new_positions

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
sorted_arr = library_sort(arr)
print("정렬된 배열:", sorted_arr[:10])
```

**Code of Tim Sort**

```
import random

def insertion_sort(arr, left, right):
    for i in range(left + 1, right + 1):
        key = arr[i]
        j = i - 1
        while j >= left and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

def merge(arr, l, m, r):
    left = arr[l:m + 1]
    right = arr[m + 1:r + 1]

    i = j = 0
    k = l

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            arr[k] = left[i]
            i += 1
        else:
            arr[k] = right[j]
            j += 1
        k += 1

    while i < len(left):
        arr[k] = left[i]
        i += 1
        k += 1

    while j < len(right):
        arr[k] = right[j]
        j += 1
        k += 1

def tim_sort(arr):
    n = len(arr)
    RUN = 32

    # insertion sort for small runs
    for i in range(0, n, RUN):
        insertion_sort(arr, i, min(i + RUN - 1, n - 1))

    # merge runs
    size = RUN
    while size < n:
        for left in range(0, n, 2 * size):
            mid = min((left + size - 1), n - 1)
            right = min((left + 2 * size - 1), n - 1)

            if mid < right:
                merge(arr, left, mid, right)

        size *= 2

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
tim_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Cocktail Shaker Sort**

```
import random

def cocktail_shaker_sort(arr):
    n = len(arr)
    start = 0
    end = n - 1
    swapped = True

    while swapped:
        swapped = False

        # left to right
        for i in range(start, end):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                swapped = True
        end -= 1

        if not swapped:
            break

        swapped = False

        # right to left
        for i in range(end, start, -1):
            if arr[i] < arr[i - 1]:
                arr[i], arr[i - 1] = arr[i - 1], arr[i]
                swapped = True
        start += 1

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
cocktail_shaker_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Comb Sort**

```
import random

def comb_sort(arr):
    n = len(arr)
    gap = n
    shrink = 1.3  # make gap
    sorted = False

    while not sorted:
        gap = int(gap / shrink)
        if gap <= 1:
            gap = 1
            sorted = True

        i = 0
        while i + gap < n:
            if arr[i] > arr[i + gap]:
                arr[i], arr[i + gap] = arr[i + gap], arr[i]
                sorted = False
            i += 1

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
comb_sort(arr)
print("정렬된 배열:", arr[:10])
```

**Code of Tournament Sort**

```
import random
import math

def build_tournament(arr):
    n = len(arr)
    size = 2 ** (math.ceil(math.log2(n)))
    tree = [float('inf')] * (2 * size)

    # fill tree
    for i in range(n):
        tree[size + i] = arr[i]

    for i in range(size - 1, 0, -1):
        tree[i] = min(tree[2 * i], tree[2 * i + 1])

    return tree, size

def remove_min(tree, size):
    index = 1
    # find minimum position
    while index < size:
        if tree[2 * index] == tree[index]:
            index = 2 * index
        else:
            index = 2 * index + 1
    tree[index] = float('inf')  # remove

    # update tree
    while index > 1:
        index //= 2
        tree[index] = min(tree[2 * index], tree[2 * index + 1])

def tournament_sort(arr):
    result = []
    tree, size = build_tournament(arr)
    for _ in range(len(arr)):
        result.append(tree[1])  # root == minimum
        remove_min(tree, size)
    return result

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
sorted_arr = tournament_sort(arr)
print("정렬된 배열:", sorted_arr[:10])
```

**Code of Intro Sort**

```
import random
import math

def insertion_sort(arr, start, end):
    for i in range(start + 1, end + 1):
        key = arr[i]
        j = i - 1
        while j >= start and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

def heapify(arr, n, i, start):
    largest = i
    l = 2 * (i - start) + 1 + start
    r = 2 * (i - start) + 2 + start

    if l < n and arr[l] > arr[largest]:
        largest = l
    if r < n and arr[r] > arr[largest]:
        largest = r
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest, start)

def heap_sort(arr, start, end):
    n = end + 1
    for i in range((start + end) // 2, start - 1, -1):
        heapify(arr, n, i, start)
    for i in range(end, start, -1):
        arr[start], arr[i] = arr[i], arr[start]
        heapify(arr, i, start, start)

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def introsort_util(arr, start, end, maxdepth):
    size = end - start + 1
    if size < 16:
        insertion_sort(arr, start, end)
    elif maxdepth == 0:
        heap_sort(arr, start, end)
    else:
        pivot = partition(arr, start, end)
        introsort_util(arr, start, pivot - 1, maxdepth - 1)
        introsort_util(arr, pivot + 1, end, maxdepth - 1)

def intro_sort(arr):
    maxdepth = int(2 * math.log2(len(arr)))
    introsort_util(arr, 0, len(arr) - 1, maxdepth)

# test
size = 10000
arr = [random.randint(0, size) for _ in range(size)]
print("원래 배열: ", arr[:10])
intro_sort(arr)
print("정렬된 배열:", arr[:10])
```
